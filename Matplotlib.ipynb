```python
import pandas as pd # for dataframe 
import numpy as np #for array
import matplotlib.pyplot as plt
```

Matplotlib basics

```python
x=np.linspace(0,5,11)
y=x**2
```

```python
x   #array([0. , 0.5, 1. , 1.5, 2. , 2.5, 3. , 3.5, 4. , 4.5, 5. ])
y   #array([ 0.  ,  0.25,  1.  ,  2.25,  4.  ,  6.25,  9.  , 12.25, 16.  , 20.25, 25.  ])
```

```python
plt.plot(x,y)
plt.title("HELLO")
plt.xlabel('x axis')
plt.ylabel('y axis')
```

subplots in matplotlib

```python
plt.subplot(2,2,1) # 2 rows and 2 columns and  1 means "a11"
plt.plot(y**2,x**5)
plt.subplot(2,2,2)
plt.plot(x**2,x**2)
plt.subplot(2,2,3)
plt.plot(y,x)
plt.subplot(2,2,4)
plt.plot(x,y)
```

```python
plt.subplot(6,6,1)# 6 row and 6 columns
plt.subplot(6,6,2)
plt.subplot(6,6,3)
plt.subplot(6,6,4)
plt.subplot(6,6,5)
plt.subplot(6,6,6)
#it means 36 rowa and columns 6*6=36
```

Object oriented way

Matplotlib allows the aspect ratio,DPI and figure size to be specified when the figure object is created,you can use the figsize and dpi keyword arguments

figsize is a tuple of the width and height of the figure in inches dpi is the dots-per-inches(pixel per inches)

```python
fig=plt.figure(figsize=(10,6),dpi=100)       #dpi means clear the image #DPI = Dots Per Inch = Image quality/detail
#fig.add_axes([0.1,0.1,0.8,0.8])#between 0 to 1
axis1=fig.add_axes([0.1,0.1,1,1])
axis1.plot(x,y)
axis2=fig.add_axes([0.3,0.4,.3,.3])
axis2.plot(x,y)


#in this full controls in canvas 
#but in sublpot we can't  control canvas , it automatically shows.
```

Types of plots

| Plot             | Main Purpose                       |
| ---------------- | ---------------------------------- |
| **Line Plot**    | Trend/change                       |
| **Bar Plot**     | Category comparison                |
| **Histogram**    | Distribution/frequency             |
| **Scatter Plot** | Relationship between 2 variables   |
| **Pie Chart**    | Percentage/parts of whole          |
| **Box Plot**     | Distribution + outliers            |
| **Area Plot**    | Trend + magnitude                  |
| **Heatmap**      | Values/correlation using intensity |

```python
fig=plt.figure()
ax=fig.add_axes([0.1,0.1,0.8,0.8])
ax.plot(x,y)
plt.savefig('basic plot.png')
```

```python
plt.figure(figsize=(8, 5), dpi=600)

plt.plot(x, y,
         color='blue',
         marker='o',
         linestyle='--',
         linewidth=2,
         label='Data')

plt.title("My Graph")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.grid(True)
plt.legend()

plt.show()
```

```python
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]

sales_A = [20, 25, 30, 28, 35, 40]
sales_B = [15, 22, 27, 32, 30, 38]
sales_C = [10, 18, 25, 20, 28, 35]

plt.figure(figsize=(10, 6), dpi=600)

plt.plot(months, sales_A,
         color="red",
         marker="o",
         linestyle="-",
         linewidth=2,
         label="Product A")

plt.plot(months, sales_B,
         color="blue",
         marker="s",
         linestyle="--",
         linewidth=2,
         label="Product B")

plt.plot(months, sales_C,
         color="green",
         marker="^",
         linestyle=":",
         linewidth=2,
         label="Product C")

plt.title("Monthly Sales Comparison", fontsize=16)
plt.xlabel("Month", fontsize=12)
plt.ylabel("Sales", fontsize=12)

plt.grid(True, linestyle="--", alpha=0.5)
plt.legend()

plt.show()
```
