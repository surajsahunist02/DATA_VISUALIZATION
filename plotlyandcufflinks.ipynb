```python
import pandas as pd
import numpy as np
import seaborn as sns
import cufflinks as cf
import plotly.express as px
import matplotlib.pyplot as plt
```

```python
tips=sns.load_dataset('tips')
```

```python
tips
```

```python
fig = px.histogram(tips, x='total_bill',y='tip')
fig.show()
```

```python
fig = px.bar(tips, x='day',y='total_bill')
fig.show()
```

```python
fig = px.scatter(tips, x='total_bill', y='tip')
fig.show()
```

```python
fig = px.box(tips, x='total_bill',y='day')
fig.show()
```

```python
fig = px.line(tips, x='total_bill',y='day')
fig.show()
```
