CATEGORICAL DATA PLOT

USING SEABORN TO PLOT CATEGORICAL DATA! There are a few main plot types for this

1.boxplot
2.violinplot
3.stripplot
4.swarmplot
5.barplot
6.countplot



```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=sns.load_dataset('tips')
```

```python
df
```

```python
sns.countplot(x = df['sex'],hue = df['smoker'])
```

```python
sns.barplot(x = df['sex'],y =  df['total_bill'])
```

```python
import numpy as np
sns.barplot(x = df['sex'],y =  df['total_bill'],estimator=np.sum)
```

```python
sns.barplot(x = df['sex'],y =  df['tip'],estimator=np.sum)
```

```python
sns.boxplot(df)
```

```python
sns.boxplot(x='tip',y='day',data=df)
```

Box PlotStructure:
Features a central box from the first quartile (25th percentile) to the third quartile (75th percentile), with a line inside for the median.Whiskers & Outliers: Lines extend out to show data variability (typically 1.5 × IQR), with individual dots representing statistical outliers.Best Use: Simple and clean for comparing central tendencies and spread across groups, especially with large numbers of categories or limited space.


Violin PlotStructure:
Combines a traditional box plot with a rotated kernel density estimation (KDE) plotted symmetrically on both sides.Density Shape: The width of the "violin" at any given point represents the data density or frequency of values—wider sections mean more data points are clustered there.Best Use: Ideal for revealing complex data patterns like multi-modal distributions (multiple peaks) that a standard box plot hides.

```python
sns.violinplot(x = 'day',y = "total_bill",data=df)
```

Strip PlotStructure:


Plots every individual data point along a single categorical axis line.Overlapping: When multiple data points share similar values, they stack directly on top of each other, creating a dense "strip" or line.Best Use: Highlighting exact individual values for small datasets, or showing overall range boundaries when paired with a box plot.

Swarm Plot (Beeswarm)Structure:


Similar to a strip plot, but it uses a specialized positioning algorithm to spread data points sideways along the categorical axis.Overlapping: Points never overlap; they sit side-by-side to form a cloud that resembles a swarm of bees.Best Use: Visualizing both individual data points and the shape of the data distribution simultaneously for small-to-medium datasets.

```python
sns.stripplot(x = 'day', y = 'total_bill',data=df)
```

```python
sns.swarmplot(x = 'day', y = "total_bill",data=df)
```

```python
sns.violinplot(x = "tip", y = 'day',data=df)
sns.swarmplot(x = "tip", y = 'day',data=df)
```
