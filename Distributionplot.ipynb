```python
sns.histplot(df['total_bill'],bins=20)#you can chnage the bins
```

```python
sns.histplot(df['total_bill'],bins=20,kde=True)#if you check the distribution of data then apply histplot
```

```python
plt.subplot(1,2,1)
sns.histplot(df['total_bill'],bins=20,kde=True)
plt.subplot(1,2,2)
sns.histplot(df['tip'],bins=20,kde=True)
```

```python
sns.jointplot(x = 'total_bill', y = 'tip', data=df,kind='scatter' )#if you check 2 distribution / quantitative data then apply jointplot
```

```python
'''
kind : { "scatter" | "kde" | "hist" | "hex" | "reg" | "resid" }
    Kind of plot to draw. See the examples for references to the underlying functions.
'''
sns.jointplot(x = 'total_bill', y = 'tip', data=df,kind='reg' )
```

```python
sns.pairplot(df,hue="sex")#hue means divide the data into categorized.
```

```python
sns.pairplot(df,hue="size",palette='rainbow')#palette means colourfull the data
```

```python
sns.rugplot(df)
```
